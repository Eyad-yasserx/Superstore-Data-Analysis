## 3. Reusable Preprocessing Functions

```python
def clean_dates(data):
    data['Order Date'] = pd.to_datetime(data['Order Date'], format='%m/%d/%Y', errors='coerce')
    data['Ship Date'] = pd.to_datetime(data['Ship Date'], format='%m/%d/%Y', errors='coerce')
    return data

def clean_duplicates(data):
    return data.drop_duplicates()

def clean_missing_values(data):
    return data.dropna(subset=['Order ID'])

def create_features(data):
    data['Profit Margin'] = (data['Profit'] / data['Sales']) * 100
    data['Shipping Duration'] = (data['Ship Date'] - data['Order Date']).dt.days
    data['Sales Performance Category'] = pd.cut(
        data['Sales'], bins=[0, 100, 500, np.inf],
        labels=['Low', 'Medium', 'High']
    )
    return data

def handle_outliers(data, columns):
    for column in columns:
        q1 = data[column].quantile(0.25)
        q3 = data[column].quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        data[column] = data[column].clip(lower=lower, upper=upper)
    return data
```

## 4. Data Cleaning

```python
df = clean_missing_values(df)
df = clean_duplicates(df)
df = clean_dates(df)
df['Postal Code'] = df['Postal Code'].astype('Int64').astype('string')
df.loc[(df['City'] == 'Burlington') & (df['State'] == 'Vermont'), 'Postal Code'] = '05401'
df = create_features(df)
```

## 17. OOP — Simple DataCleaner Class

```python
class DataCleaner:
    def __init__(self, data):
        self.data = data.copy()

    def optimize_types(self):
        self.data['Row ID'] = self.data['Row ID'].astype('Int32')
        self.data['Quantity'] = self.data['Quantity'].astype('int16')
        self.data['Sales'] = self.data['Sales'].astype('float32')
        self.data['Discount'] = self.data['Discount'].astype('float32')
        self.data['Profit'] = self.data['Profit'].astype('float32')
        return self.data
```

```python
cleaner = DataCleaner(df)
df = cleaner.optimize_types()
df.dtypes
```

## 18. KPI Summary

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
average_order_value = total_sales / total_orders
average_profit_margin = df['Profit Margin'].mean()
average_shipping_duration = df['Shipping Duration'].mean()

kpi_summary = {
    'Total Sales': total_sales,
    'Total Profit': total_profit,
    'Total Orders': total_orders,
    'Average Order Value': average_order_value,
    'Average Profit Margin': average_profit_margin,
    'Average Shipping Duration': average_shipping_duration
}
kpi_summary
```

## 19. Automated Analytical Summary

```python
analytical_summary = {
    'Highest Sales Category': df.groupby('Category')['Sales'].sum().idxmax(),
    'Highest Profit Category': df.groupby('Category')['Profit'].sum().idxmax(),
    'Highest Sales Region': df.groupby('Region')['Sales'].sum().idxmax(),
    'Highest Sales Segment': df.groupby('Segment')['Sales'].sum().idxmax(),
    'Average Shipping Duration': df['Shipping Duration'].mean(),
    'Average Profit Margin': df['Profit Margin'].mean()
}
for key, value in analytical_summary.items():
    print(f'{key}: {value}')
```
